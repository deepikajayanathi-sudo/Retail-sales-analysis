```python
import pandas as pd
import numpy as np

#step 1 reading and exploring csv file
df = pd.read_csv("C:/Users/ADMIN/Documents/Retail sales analytics/Dataset/retail_dataset.csv")
print(df.head())
print(df.info())
print(df.describe())
print(df.columns)
print(df.shape)
print(df.sample(5))
print(df.dtypes)
```

```python
#STEP 2: HANDLING MISSING VALUES
print("missing values in each column:")
print(df.isnull().sum())
#convert in numeric values
df["Quantity"] = pd.to_numeric(df["Quantity"], errors='coerce')
df["Price per Unit"] = pd.to_numeric(df["Price per Unit"], errors='coerce')
df["Total Amount"] = pd.to_numeric(df["Total Amount"], errors='coerce')

#fill numeric missing values
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Price per Unit"] = df["Price per Unit"].fillna(df["Price per Unit"].median())

#fill categorical missing values
df["Age"] = df["Age"].fillna("unknown")
df["Gender"] = df["Gender"].fillna("unknown")

#drop rows where Total Amount is missing
df = df.dropna(subset=["Total Amount"])


```

```python
#step:3 handling duplicate rows in csv file

print("Number of duplicate rows:", df.duplicated().sum())
#remove duplicate rows  
df = df.drop_duplicates()
print("Number of duplicate rows after removal:", df.duplicated().sum())

```

```python
#STEP:4 DEALING WITH OUTLIERS
df=df.reset_index(drop=True)
# CALCULATE Q1 AND Q3
Q1 = df["Total Amount"].quantile(0.25)
Q3 = df["Total Amount"].quantile(0.75)
#FIND IQR
IQR = Q3 - Q1

#DEFINE LOWER AND UPPER BOUNDS
LOWER_BOUND = Q1 - 1.5 * IQR
UPPER_BOUND = Q3 + 1.5 * IQR

#REMOVE OUTLIERS
df = df[(df["Total Amount"] >= LOWER_BOUND) & (df["Total Amount"] <= UPPER_BOUND)]

print("outliers removed successfully")


```

```python
#step:5 data type conversion

#convert in numeric values
df["Quantity"] = pd.to_numeric(df["Quantity"], errors='coerce')
df["Price per Unit"] = pd.to_numeric(df["Price per Unit"], errors='coerce')
df["Total Amount"] = pd.to_numeric(df["Total Amount"], errors='coerce')

#convert data colunm (if available)
df["Date"] = pd.to_datetime(df["Date"], errors='coerce')
print("data type after conversion:")
print(df.dtypes)
```

```python
#step:6 feature engineering

#create a new colunm :average price
df["Average Price"] = df["Total Amount"] / df["Quantity"].replace(0, 1)

# create a new colunm : total items values(optional examples)
df["Total Items Value"] = df["Quantity"] * df["Price per Unit"]

print("feature engineering completed successfully")
print(df.head())
```

```python
#step:7 encoding categorical variables

#encoding categorical variables using one-hot encoding
df=pd.get_dummies(df, columns=["Gender", "Age"], drop_first=True)

print("categorical variables encoded successfully")
print(df.head())
```

```python
# Step 8: Date and Time Manipulation

# convert Date column into datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# extract useful date features
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day

print("Date features extracted successfully")
print(df.head())

```

```python
# Step 9: Data Validation and Consistency Check

# calculate expected total amount
df["Calculated Total"] = df["Quantity"] * df["Price per Unit"]

# check whether Total Amount is correct
df["Amount Check"] = df["Total Amount"] == df["Calculated Total"]

print("Inconsistent records:")
print(df[df["Amount Check"] == False])

# remove temporary columns
df = df.drop(columns=["Calculated Total", "Amount Check"])

print("Data validation completed successfully")
```

```python
import os

os.makedirs("data", exist_ok=True)

df.to_csv("data/final_cleaned_sales.csv", index=False)

print("Final cleaned dataset saved successfully")
print("Final shape:", df.shape)
```
